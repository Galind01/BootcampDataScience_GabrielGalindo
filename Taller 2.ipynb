```python
import pandas as pd
import numpy as np
pizza_data = pd.read_excel('Enhanced_pizza_sell_data_2024-25.xlsx')
pizza_data.head()
```

```python
pizza_data.columns
```

```python
#1. Orden de mayor a menor ventas los nombres de los restaurantes
print("--- Restaurantes por Ventas (Mayor a Menor) ---")
restaurante_ventas = pizza_data['Restaurant Name'].value_counts()
print(restaurante_ventas)
print("\n")
```

```python

#2. Que estadisticas maneja cada compra en el local?
print("Que estadisticas maneja cada compra en el local?")
# Seleccionar columnas numéricas para estadísticas descriptivas, excluyendo identificadores únicos si los hubiera
columnas_numericas = pizza_data.select_dtypes(include=np.number).columns
if 'Order ID' in columnas_numericas:
    columnas_numericas = columnas_numericas.drop('Order ID')

if len(columnas_numericas) > 0:
    estadisticas_descriptivas = pizza_data[columnas_numericas].describe()
    print(estadisticas_descriptivas)
else:
    print("No se encontraron columnas numéricas adecuadas para estadísticas descriptivas.")
print("\n")
```

```python
#3. Orden de mayor a menor ventas de los nombres y tipos de pizza
print("--- Tipos de Pizza por Ventas (Mayor a Menor) ---")
# Combinar nombre y tipo de pizza para un análisis más detallado
pizza_data['Pizza Description'] = pizza_data['Pizza Type'] + ' - ' + pizza_data['Pizza Type'] # Asumiendo que 'Pizza Type' es la columna relevante para el tipo

# Contar las ventas por tipo de pizza
tipo_pizza_ventas = pizza_data['Pizza Description'].value_counts()
print(tipo_pizza_ventas)
print("\n")
```
