```python

def calcular_ingreso(precio, cantidad):
    """
    Calcula el ingreso total multiplicando el precio por la cantidad.

    Args:
        precio (float): El precio unitario del Producto 
        cantidad (int): La cantidad de productos.

    Returns:
        float: El ingreso total calculado, o 0 si ocurre un error.
   """
    try:
        ingreso_total = float(precio) * int(cantidad)
        return ingreso_total
    except (ValueError, TypeError):
        print("Error: Los valores deben ser numéricos")
        return 0




```
