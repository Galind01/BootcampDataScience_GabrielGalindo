```python
import csv
import requests

# 1. Obtener los arquetipos
url_arquetipos = "https://db.ygoprodeck.com/api/v7/archetypes.php"
res_arquetipos = requests.get(url_arquetipos).json()

# Seleccionar la cantidad deseada (por ejemplo, los primeros 10)
arquetipos = [item["archetype_name"] for item in res_arquetipos[:]]

archivo_csv = "mazos_yugioh_completo.csv"

with open(archivo_csv, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    # Definir columnas
    writer.writerow(["Arquetipo", "ID_Carta", "Nombre_Carta", "Tipo", "Raza", "Atributo", "ATK", "DEF"])

    for arquetipo in arquetipos:
        url_cartas = f"https://db.ygoprodeck.com/api/v7/cardinfo.php?archetype={requests.utils.quote(arquetipo)}"
        res_cartas = requests.get(url_cartas)
        
        if res_cartas.status_code == 200:
            cartas = res_cartas.json().get("data", [])
            for c in cartas:
                writer.writerow([
                    arquetipo,
                    c.get("id"),
                    c.get("name"),
                    c.get("type"),
                    c.get("race", ""),
                    c.get("attribute", "N/A"),
                    c.get("atk", "N/A"),
                    c.get("def", "N/A")
                ])

print(f"Archivo '{archivo_csv}' generado exitosamente.")
```
